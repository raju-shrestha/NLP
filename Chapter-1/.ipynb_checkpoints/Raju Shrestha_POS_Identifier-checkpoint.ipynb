```python
# Import Required Libraries
import nltk
import pandas as pd
from nltk.tokenize import word_tokenize
from nltk import pos_tag
```

```python
# Download Required NLTK Resources
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger_eng')
```

### News Article Link:
https://thehimalayantimes.com/opinion/43rd-b-p-memorial-day-literary-contribution-superseding-the-political

```python
# Load the News Article
with open("news_article.txt", "r", encoding="utf-8") as file:
    text = file.read()

print(text[:])   # Display first 500 characters
```

```python
# Tokenize the Text
words = word_tokenize(text)

print(words[:])
```

```python
# Perform POS Tagging
tagged_words = pos_tag(words)

print(tagged_words[:])
```

```python
# Extract Nouns and Verbs
noun_tags = ['NN', 'NNS', 'NNP', 'NNPS']
verb_tags = ['VB', 'VBD', 'VBG', 'VBN', 'VBP', 'VBZ']

extracted = []

for word, tag in tagged_words:
    if tag in noun_tags or tag in verb_tags:
        extracted.append([word, tag])

print(extracted[:])
```

```python
# Create a DataFrame
df = pd.DataFrame(extracted, columns=["Word", "POS_Tag"])
df.head()
```

```python
# Save the Result to CSV File
df.to_csv("Extracted_Nouns_Verbs.csv", index=False)

print("CSV file saved successfully.")
```
