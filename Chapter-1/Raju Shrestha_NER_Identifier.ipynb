```python
# Import Required Libraries
import spacy
import pandas as pd
from collections import Counter
```

```python
# Load the English Language Model
nlp = spacy.load("en_core_web_sm")
```

### News Article Link:
https://thehimalayantimes.com/opinion/43rd-b-p-memorial-day-literary-contribution-superseding-the-political

```python
# Load the News Article
with open("news_article.txt", "r", encoding="utf-8") as file:
    text = file.read()

print(text[:])
```

```python
# Perform Named Entity Recognition
doc = nlp(text)
```

```python
# Extract Named Entities
entities = []

for ent in doc.ents:
    entities.append([ent.text, ent.label_])

df = pd.DataFrame(entities, columns=["Entity", "Entity_Type"])

df.head(20)
```

```python
# Save the Results to CSV
df.to_csv("Named_Entities.csv", index=False)

print("CSV file saved successfully.")
```

```python
# Summary of Entity Types
entity_summary = df["Entity_Type"].value_counts()

print(entity_summary)
```
