### Exercise 1: Word Frequency Counter
* Objective: Count how many times each word appears in a sentence.

### Method 1: Using Dictionary

```python
# Input text
text = "I like NLP I like Python NLP is fun"

# Convert to lowercase
text = text.lower()

# Split sentence into words
words = text.split()

# Dictionary to store word frequencies
frequency = {}

for word in words:
    if word in frequency:
        frequency[word] += 1
    else:
        frequency[word] = 1

# Display result
print("\nWord Frequencies:")
for word, count in frequency.items():
    print(word, ":", count)
```

### Exercise 3: Build a Word Co-occurrence Matrix
* Objective: Construct a simple Word Co-occurrence Matrix using a window size of 1.

```python
import pandas as pd

# Input sentence
sentence = "the cat sat on the mat"

# Split sentence into words
words = sentence.split()

# Build vocabulary (unique words)
vocabulary = sorted(set(words))

# Create word-to-index mapping
word_index = {word: i for i, word in enumerate(vocabulary)}

# Create an empty matrix
matrix = [[0 for _ in vocabulary] for _ in vocabulary]

# Window size = 1
window_size = 1

# Count neighboring words
for i, word in enumerate(words):

    # Left neighbor
    if i - window_size >= 0:
        neighbor = words[i - window_size]
        matrix[word_index[word]][word_index[neighbor]] += 1

    # Right neighbor
    if i + window_size < len(words):
        neighbor = words[i + window_size]
        matrix[word_index[word]][word_index[neighbor]] += 1

# Display as DataFrame
df = pd.DataFrame(matrix, index=vocabulary, columns=vocabulary)

print(df)
```

### Exercise 4: Detect Data Sparsity
* Objective: Detect word pairs that never occur together using the Word Co-occurrence Matrix created in Exercise 3.

```python
import pandas as pd

# Input sentence
sentence = "the cat sat on the mat"
print("Input sentence: ", sentence)

# Split into words
words = sentence.split()

# Build vocabulary
vocabulary = sorted(set(words))

# Create word-to-index mapping
word_index = {word: i for i, word in enumerate(vocabulary)}

# Create empty co-occurrence matrix --> Creates a matrix initialized with zeros.
matrix = [[0 for _ in vocabulary] for _ in vocabulary]

# Window size = 1
for i, word in enumerate(words):

    if i > 0:
        left = words[i - 1]
        matrix[word_index[word]][word_index[left]] += 1   # Count Neighboring Words

    if i < len(words) - 1:
        right = words[i + 1]
        matrix[word_index[word]][word_index[right]] += 1

# Display Co-occurrence Matrix
df = pd.DataFrame(matrix, index=vocabulary, columns=vocabulary)

print("Co-occurrence Matrix:\n")
print(df)

print("\nWord pairs that never occurred together:\n")

# Find zero-valued cells (excluding diagonal)
for i in range(len(vocabulary)):       # Checks every cell in the matrix
    for j in range(len(vocabulary)):
        if i != j and matrix[i][j] == 0:    # Detect Sparsity, the two words never occurred together.
            print(f"({vocabulary[i]}, {vocabulary[j]}) -> Never occurred")
```

### Exercise 5: Implement Laplace Smoothing
* Objective: Calculate word probabilities using Laplace Smoothing (Add-One Smoothing).

```python
# Input word counts --> A dictionary stores each word and its frequency
word_counts = {
    "cat": 4,
    "dog": 3,
    "bird": 1
}

# Compute total words --> Adds all word frequencies
total_words = sum(word_counts.values())
print("Total Words =", total_words)

# Compute vocabulary size --> Counts the number of unique words.
vocabulary_size = len(word_counts)
print("Vocabulary Size =", vocabulary_size)


print("\nLaplace Smoothed Probabilities:\n")
# Apply Laplace Smoothing
for word, count in word_counts.items():
    probability = (count + 1) / (total_words + vocabulary_size)   # Adds 1 to each word count before calculating the probability.
    print(f"{word} : {probability:.4f}")   # Displays the probability rounded to four decimal places.
```

### Exercise 6: Simple Keyboard Prediction
* Objective: Predict the most common next word based on the given training corpus using a Bigram Language Model.

```python
# Training corpus
corpus = [
    "I like NLP",
    "I like Python",
    "I like coffee",
    "I love NLP"
]

# Dictionary to store bigrams
bigrams = {}

# Build bigram counts
for sentence in corpus:
    words = sentence.split()

    for i in range(len(words) - 1):
        current = words[i]
        next_word = words[i + 1]

        if current not in bigrams:
            bigrams[current] = {}

        if next_word not in bigrams[current]:
            bigrams[current][next_word] = 1
        else:
            bigrams[current][next_word] += 1

# Input
current_word = "like"

# Display predictions
print("Suggestions:")

if current_word in bigrams:
    # Sort by highest frequency
    predictions = sorted(bigrams[current_word].items(),
                         key=lambda x: x[1],
                         reverse=True)

    for word, count in predictions[:3]:
        print(word)
else:
    print("No prediction found")
```

### Exercise 7: Calculate TF-IDF
* Objective: Compute Term Frequency (TF), Document Frequency (DF), Inverse Document Frequency (IDF), and TF-IDF scores for the given corpus.
* Tasks
    * Calculate TF.
    * Calculate DF.
    * Calculate IDF.
    * Compute TF-IDF.

```python
import math

# Documents
documents = [
    "I like NLP",
    "I like AI",
    "AI is amazing"
]

# ---------- TF ----------
tf = []

for doc in documents:
    words = doc.split()
    total = len(words)

    freq = {}
    for word in words:
        freq[word] = words.count(word) / total

    tf.append(freq)

print("Term Frequency")
for i in range(len(tf)):
    print("Document", i + 1, ":", tf[i])

print("=" * 50)

# ---------- DF ----------
df = {}

for doc in documents:
    for word in set(doc.split()):
        if word in df:
            df[word] += 1
        else:
            df[word] = 1

print("Document Frequency")
print(df)

print("=" * 50)

# ---------- IDF ----------
idf = {}
N = len(documents)

for word in df:
    idf[word] = math.log10(N / df[word])

print("Inverse Document Frequency")
print(idf)

print("=" * 50)

# ---------- TF-IDF ----------
print("TF-IDF")
for i in range(len(tf)):
    print("\nDocument", i + 1)

    for word in tf[i]:
        print(word, ":", round(tf[i][word] * idf[word], 4))
```

### Exercise 8: Dense Vector Similarity
* Objective: Compute cosine similarity between two dense vectors.
* Input:
    * cat = [0.4,0.2,0.8]
    * dog = [0.5,0.1,0.7]
* Expected Output:
* Cosine Similarity = 0.98(Approximate value.)
* Tasks:
    * Compute dot product.
    * Compute vector magnitudes.
    * Calculate cosine similarity.
* Formula:


```python
import math # We use the math library to calculate square roots with math.sqrt()

# Dense vectors
cat = [0.4, 0.2, 0.8]  # creates a dense vector for the word cat
dog = [0.5, 0.1, 0.7]  # creates a dense vector for the word dog

# 1. Compute dot product --> calculates the dot product of the two vectors
dot_product = sum(c * d for c, d in zip(cat, dog))   # Multiply each pair

# 2. Compute vector magnitudes
magnitude_cat = math.sqrt(sum(c ** 2 for c in cat)) # Calculate magnitude of cat
magnitude_dog = math.sqrt(sum(d ** 2 for d in dog)) # Calculate magnitude of dog

# 3. Calculate cosine similarity
cosine_similarity = dot_product / (magnitude_cat * magnitude_dog) # Calculate cosine similarity

print("Dot Product =", dot_product)
print("Magnitude of cat =", magnitude_cat)
print("Magnitude of dog =", magnitude_dog)
print("Cosine Similarity =", round(cosine_similarity, 2))
```

### Exercise 9: Find Similar Words
* Objective: Find the word most similar to a given word.
* Tasks
    * Compare the target vector with all others.
    * Compute cosine similarity.
    * Print the most similar word.


```python
import math   # We use the math library to calculate square roots with math.sqrt()

# Create Word embeddings --> dictionary contains a vector for each word
embeddings = {
    "cat": [0.4, 0.2, 0.8],
    "dog": [0.5, 0.1, 0.7],
    "car": [-0.7, 0.8, -0.2],
    "lion": [0.45, 0.25, 0.82]
}

# Select the target word
target_word = "cat"  # We want to find which word is most similar to cat

# Get the vector of the target word
target_vector = embeddings[target_word]

# Function to calculate cosine similarity
def cosine_similarity(vector1, vector2):
    dot_product = sum(a * b for a, b in zip(vector1, vector2)) # Calculate dot product, and It multiplies correspondng values and adds them

    # Calculate magnitude
    magnitude1 = math.sqrt(sum(a ** 2 for a in vector1)) 
    magnitude2 = math.sqrt(sum(b ** 2 for b in vector2))

    # Calculate cosine similarity
    return dot_product / (magnitude1 * magnitude2)

# Store similarities
similarities = {}  # Store the similarity of cat with each other word

# Compare cat with all other words
for word, vector in embeddings.items():

    # Skip cat itself
    if word != target_word:  # We don't compare cat with itself because: Similarity(cat,cat)=1
        similarity = cosine_similarity(target_vector, vector) # Calculate similarity
        similarities[word] = similarity # Store the result

# Find the most similar word
most_similar_word = max(similarities, key=similarities.get) # max() finds the word with the largest similarity value

# Print results
print("Most similar to", target_word)
print(most_similar_word)
print("Similarity =", round(similarities[most_similar_word], 2))
```

### Exercise 10 (Assignment) : Mini NLP Pipeline
* Objective: Combine multiple concepts into one application.
* Tasks: 
1. Build the vocabulary.
2. Count word frequencies.
3. Compute TF.
4. Build the co-occurrence matrix.
5. Detect sparse entries.
6. Apply Laplace smoothing to unigram probabilities.
7. Predict the next word after a user-entered word.
8. Display the top three most frequent words.

```python
import re
import math
from collections import Counter

# 1. Input Corpus
corpus = [
    "I love NLP",
    "I love AI",
    "AI loves Python",
    "Python loves data"
]

# Convert text to lowercase and tokenize
sentences = [sentence.lower().split() for sentence in corpus]

# Flatten all words into one list
all_words = [word for sentence in sentences for word in sentence]

# 2. Build Vocabulary
vocabulary = sorted(set(all_words))
print("1. Vocabulary:")
print(vocabulary)
print("Vocabulary Size =", len(vocabulary))

# 3. Count Word Frequencies
word_frequency = Counter(all_words)

print("\n2. Word Frequencies:")

for word, count in word_frequency.items():
    print(word, ":", count)

# 4. Compute TF
print("\n3. Term Frequency (TF):")

for i, sentence in enumerate(sentences):
    counts = Counter(sentence)
    total_words = len(sentence)

    print("Sentence", i + 1)

    for word in vocabulary:
        tf = counts[word] / total_words
        if tf > 0:
            print(word, ":", round(tf, 2))

# 5. Build Co-occurrence Matrix
print("\n4. Co-occurrence Matrix:")

window_size = 1

co_occurrence = {
    word: {other: 0 for other in vocabulary}
    for word in vocabulary
}

for sentence in sentences:
    for i, word in enumerate(sentence):

        start = max(0, i - window_size)
        end = min(len(sentence), i + window_size + 1)

        for j in range(start, end):
            if i != j:
                other_word = sentence[j]
                co_occurrence[word][other_word] += 1

# Print matrix header
print("      ", end="")

for word in vocabulary:
    print(f"{word:>8}", end="")

print()

# Print matrix rows
for word in vocabulary:
    print(f"{word:>6}", end="")

    for other in vocabulary:
        print(f"{co_occurrence[word][other]:>8}", end="")

    print()

# 6. Detect Sparse Entries
print("\n5. Sparse Entries:")

sparse_entries = []

for word in vocabulary:
    for other in vocabulary:
        if co_occurrence[word][other] == 0:
            sparse_entries.append((word, other))

print("Number of zero entries =", len(sparse_entries))

print("Some sparse entries:")

for pair in sparse_entries[:10]:
    print(pair)

# 7. Laplace Smoothing
print("\n6. Laplace-Smoothed Unigram Probabilities:")

total_words = len(all_words)
vocabulary_size = len(vocabulary)

alpha = 1

for word in vocabulary:

    count = word_frequency[word]

    probability = (count + alpha) / (
        total_words + alpha * vocabulary_size
    )

    print(word, ":", round(probability, 4))


# 8. Predict Next Word
print("\n7. Next Word Prediction:")

user_word = input("Enter a word: ").lower()

next_word_counts = Counter()

for sentence in sentences:
    for i in range(len(sentence) - 1):

        if sentence[i] == user_word:
            next_word = sentence[i + 1]
            next_word_counts[next_word] += 1

if next_word_counts:
    predicted_word = next_word_counts.most_common(1)[0][0]

    print("Predicted next word =", predicted_word)

else:
    print("Word not found or no next word available.")

# 9. Top Three Most Frequent Words
print("\n8. Top Three Most Frequent Words:")

top_three = word_frequency.most_common(3)

for word, count in top_three:
    print(word, ":", count)
```

### Exercise 11: Predict the Next Word Using Conditional Probability
* Objective: Use conditional probability to predict the most likely next word after a given word using a
small corpus.
* Background: 

* Programming Task:
* Write a Python program that:
    1. Reads the training corpus.
    2. Builds bigram counts.
    3. Calculates conditional probabilities for every bigram.
    4. Asks the user to enter a word.
    5. Displays all possible next words and their conditional probabilities.
    6. Predicts the word with the highest conditional probability.

```python
from collections import Counter  # Counter is used to count how many times words and bigrams occur
# 1. Training Corpus
corpus = [
    "I like NLP",
    "I like Python",
    "I like coffee",
    "I love NLP",
    "You like coffee"
]

# Convert sentences to lowercase and split into words
sentences = [sentence.lower().split() for sentence in corpus]


# 2. Build Bigram Counts
bigram_counts = Counter()  # Stores pairs of consecutive words
word_counts = Counter()    # Storespairs of consecutive words

for sentence in sentences:  # Processes each sentence one at a time

    for i in range(len(sentence) - 1):  # Loop through words

        # Get the two words
        word1 = sentence[i]
        word2 = sentence[i + 1]

        # Count the bigram
        bigram_counts[(word1, word2)] += 1 # Every time the same bigram appears, its count increases

        # Count word1 when it appears as the first word of a bigram
        word_counts[word1] += 1

# 3. Calculate Conditional Probabilities
conditional_probabilities = {}

for (word1, word2), count in bigram_counts.items():  # Calculate conditional probability

    probability = count / word_counts[word1]   # Calculate probability

    conditional_probabilities[(word1, word2)] = probability  # Store the probability

# 4. Display Bigram Counts
print("BIGRAM COUNTS:")
print("----------------")

for (word1, word2), count in bigram_counts.items():
    print(word1, "->", word2, ":", count)

# 5. Display Conditional Probabilities
print("\nCONDITIONAL PROBABILITIES:")
print("--------------------------")

for (word1, word2), probability in conditional_probabilities.items():

    print("P(" + word2 + " | " + word1 + ") =",round(probability, 2))

# 6. Ask User for a Word
user_word = input("\nEnter a word: ").lower()

# 7. Find Possible Next Words
next_words = {}  # Store the possible next words

for (word1, word2), probability in conditional_probabilities.items():

    if word1 == user_word: # Check whether the first word matches
        next_words[word2] = probability  # Store possible next words


# 8. Display Next Words
if next_words:  # If the word exists in the corpus, display the possible next words

    print("\nPossible next words:")

    for word, probability in next_words.items():

        print(word, "->", round(probability, 2))

# 9. Predict Most Likely Next Word
    predicted_word = max(next_words,key=next_words.get)  # max() finds the word with the highest probability

    print("\nPredicted next word =", predicted_word)

else:

    print("\nNo next word found for:", user_word)
```
