### Exercise 1: Word Frequency Counter
* Objective: Count how many times each word appears in a sentence.

### Method 1: Using Dictionary

```python
# Input sentence
sentence = "I like NLP I like Python NLP is fun"

# Convert to lowercase
sentence = sentence.lower()

# Split sentence into words
words = sentence.split()

# Dictionary to store word frequencies
frequency = {}

for word in words:
    if word in frequency:
        frequency[word] += 1
    else:
        frequency[word] = 1

# Display result
print("\nWord Frequencies:")
for word, count in frequency.items():
    print(word, ":", count)
```

### Method 2: Using Counter

* Link for document.txt file: https://thehimalayantimes.com/nepal/nepals-tiger-population-climbs-to-429-nearly-quadrupling-since-2010

```python
from collections import Counter
import pandas as pd
import re        # Import regular expression

# Load text file
with open("document.txt", "r", encoding="utf-8") as file:
    text = file.read()

# Convert to lowercase
text = text.lower()

# Remove punctuation -->  removes all punctation while letters, numbers, spaces, and underscores are kept.
text = re.sub(r'[^\w\s]', '', text)   # matches all punctuation and special characters, and replaces them with an empty string.

# Split into words
words = text.split()

# Count word frequencies
word_frequency = Counter(words)

# Display word frequencies
print("Word Frequencies:\n")
for word, count in word_frequency.items():
    print(f"{word}: {count}")

# Save to CSV
df = pd.DataFrame(word_frequency.items(), columns=["Word", "Frequency"])
df.to_csv("Word_Frequency.csv", index=False)

print("\nWord frequencies saved to 'Word_Frequency.csv'")
```

### Exercise 2: Vocabulary Builder
* Objective: Find all unique words in a corpus.

```python
# Input corpus
corpus = [
    "I like NLP",
    "I like AI",
    "AI is amazing"
]

# Create an empty set  --> A set stores only unique values
vocabulary = set()

# Read each sentence
for sentence in corpus:

    # Split sentence into words
    words = sentence.split()

    # Add words to the set --> Add Words to the Vocabulary
    for word in words:
        vocabulary.add(word)

# Display vocabulary
print("Vocabulary:\n")

for word in sorted(vocabulary):  # Displays the words in alphabetical order
    print(word)

# Display vocabulary size
print("\nVocabulary Size =", len(vocabulary))
```

### Exercise 3: Build a Word Co-occurrence Matrix
* Objective: Construct a simple Word Co-occurrence Matrix using a window size of 1.

```python
import pandas as pd

# Input sentence
sentence = "the cat sat on the mat"

# Split sentence into words
words = sentence.split()

# Build vocabulary (unique words)
vocabulary = sorted(set(words))

# Create word-to-index mapping
word_index = {word: i for i, word in enumerate(vocabulary)}

# Create an empty matrix
matrix = [[0 for _ in vocabulary] for _ in vocabulary]

# Window size = 1
window_size = 1

# Count neighboring words
for i, word in enumerate(words):

    # Left neighbor
    if i - window_size >= 0:
        neighbor = words[i - window_size]
        matrix[word_index[word]][word_index[neighbor]] += 1

    # Right neighbor
    if i + window_size < len(words):
        neighbor = words[i + window_size]
        matrix[word_index[word]][word_index[neighbor]] += 1

# Display as DataFrame
df = pd.DataFrame(matrix, index=vocabulary, columns=vocabulary)

print(df)
```

### Exercise 4: Detect Data Sparsity
* Objective: Detect word pairs that never occur together using the Word Co-occurrence Matrix created in Exercise 3.

```python
import pandas as pd

# Input sentence
sentence = "the cat sat on the mat"
print("Input sentence: ", sentence)

# Split into words
words = sentence.split()

# Build vocabulary
vocabulary = sorted(set(words))

# Create word-to-index mapping
word_index = {word: i for i, word in enumerate(vocabulary)}

# Create empty co-occurrence matrix --> Creates a matrix initialized with zeros.
matrix = [[0 for _ in vocabulary] for _ in vocabulary]

# Window size = 1
for i, word in enumerate(words):

    if i > 0:
        left = words[i - 1]
        matrix[word_index[word]][word_index[left]] += 1   # Count Neighboring Words

    if i < len(words) - 1:
        right = words[i + 1]
        matrix[word_index[word]][word_index[right]] += 1

# Display Co-occurrence Matrix
df = pd.DataFrame(matrix, index=vocabulary, columns=vocabulary)

print("Co-occurrence Matrix:\n")
print(df)

print("\nWord pairs that never occurred:\n")

# Find zero-valued cells (excluding diagonal)
for i in range(len(vocabulary)):       # Checks every cell in the matrix
    for j in range(len(vocabulary)):
        if i != j and matrix[i][j] == 0:    # Detect Sparsity, the two words never occurred together.
            print(f"({vocabulary[i]}, {vocabulary[j]}) -> Never occurred")
```

### Exercise 5: Implement Laplace Smoothing
* Objective: Calculate word probabilities using Laplace Smoothing (Add-One Smoothing).

```python
# Input word counts --> A dictionary stores each word and its frequency
word_counts = {
    "cat": 4,
    "dog": 3,
    "bird": 1
}

# Compute total words --> Adds all word frequencies
total_words = sum(word_counts.values())

# Compute vocabulary size --> Counts the number of unique words.
vocabulary_size = len(word_counts)

# Denominator
denominator = total_words + vocabulary_size

print("Total Words =", total_words)
print("Vocabulary Size =", vocabulary_size)
print("\nLaplace Smoothed Probabilities:\n")

# Apply Laplace Smoothing
for word, count in word_counts.items():
    probability = (count + 1) / denominator   # Adds 1 to each word count before calculating the probability.
    print(f"{word} : {probability:.4f}")   # Displays the probability rounded to four decimal places.
```

### Exercise 6: Simple Keyboard Prediction
* Objective: Predict the most common next word based on the given training corpus using a Bigram Language Model.

```python
from collections import defaultdict, Counter

# Training corpus
corpus = [
    "I like NLP",
    "I like Python",
    "I like coffee",
    "I love NLP"
]

# Dictionary to store bigram counts
bigram_dict = defaultdict(Counter)   # defaultdict automatically creates a dictionary when a new key is encountered.Counter counts the frequency of words.

# Build bigrams  --> Create a Bigram Dictionary
for sentence in corpus:
    words = sentence.split()  # Splits the sentence into individual words.

    for i in range(len(words) - 1):
        current_word = words[i]    # Creates a pair of consecutive words.
        next_word = words[i + 1]

        bigram_dict[current_word][next_word] += 1   # Increases the frequency count of the next word.

# Input phrase  --> User input
input_text = "I like"  

# Last word determines prediction
last_word = input_text.split()[-1]   # Extracts the last word.

print("Suggestions:\n")

if last_word in bigram_dict:
    predictions = bigram_dict[last_word].most_common(3)  # Finds the three most frequent words that follow the last word.

    for word, count in predictions:
        print(word)
else:
    print("No prediction available.")
```

### Exercise 7: Calculate TF-IDF
* Objective: Compute Term Frequency (TF), Document Frequency (DF), Inverse Document Frequency (IDF), and TF-IDF scores for the given corpus.

```python
import math

# Corpus
documents = [
    "I like NLP",
    "I like AI",
    "AI is amazing"
]

# Convert documents into word lists
docs = [doc.lower().split() for doc in documents]

# Vocabulary
vocabulary = sorted(set(word for doc in docs for word in doc))

N = len(docs)

# -------------------------------
# Calculate Document Frequency (DF)
# -------------------------------
df = {}

for word in vocabulary:
    df[word] = sum(word in doc for doc in docs)   # Counts the number of documents containing each word.

# -------------------------------
# Calculate IDF
# -------------------------------
idf = {}

for word in vocabulary:
    idf[word] = math.log(N / df[word])   # Assigns higher values to words that appear in fewer documents

# -------------------------------
# Calculate TF and TF-IDF
# -------------------------------
for i, doc in enumerate(docs):

    print(f"\nDocument {i+1}")

    total_words = len(doc)

    for word in doc:

        tf = doc.count(word) / total_words  # Computes how frequently a word appears in the current document.

        tfidf = tf * idf[word]   # Combines TF and IDF to measure the importance of a word in a document.

        print(f"{word:10} TF={tf:.3f}  IDF={idf[word]:.3f}  TF-IDF={tfidf:.3f}")
```
