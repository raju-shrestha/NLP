### Exercise 1: Word Frequency Counter
* Objective: Count how many times each word appears in a sentence.

### Method 1: Using Dictionary

```python
# Input text
text = "I like NLP I like Python NLP is fun"

# Convert to lowercase
text = text.lower()

# Split sentence into words
words = text.split()   # Divides the sentence into individual words

# Dictionary to store word frequencies
frequency = {}  # Key = word, Value = number of times the word appears

for word in words:   # Loops through each word in the words
    if word in frequency:  # Checks whether the current word is already present in the dictionary
        frequency[word] += 1   # If the word already exists, increase its count by 1
    else:               # Executes when the word is not already in the dictionary
        frequency[word] = 1  # Adds the new word to the dictionary

# Display result
print("\nWord Frequencies:")
for word, count in frequency.items():
    print(word, ":", count)
```

### Exercise 3: Build a Word Co-occurrence Matrix
* Objective: Construct a simple Word Co-occurrence Matrix using a window size of 1.

```python
import pandas as pd

# Input sentence
sentence = "the cat sat on the mat"

# Split sentence into words
words = sentence.split()

# Build vocabulary (unique words)
vocabulary = sorted(set(words))  # Removes duplicate words and sorts the words alphabetically

# Create word-to-index mapping
word_index = {word: i for i, word in enumerate(vocabulary)}  # Creates a dictionary that assigns each word an index

# Create an empty matrix
matrix = [[0 for _ in vocabulary] for _ in vocabulary]   # Creates a 5 × 5 matrix filled with zeros

# Window size = 1
window_size = 1    # one word on the left or one word on the right

# Count neighboring words
for i, word in enumerate(words):  # Loops through every word in the sentence

    # Left neighbor
    if i - window_size >= 0:       # Checks whether a left neighbor exists
        neighbor = words[i - window_size]       # Gets the left neighbor
        matrix[word_index[word]][word_index[neighbor]] += 1   # Increases the count in the matrix

    # Right neighbor
    if i + window_size < len(words):   # Checks whether a right neighbor exists
        neighbor = words[i + window_size]   # Gets the right neighbor
        matrix[word_index[word]][word_index[neighbor]] += 1    # Updates the matrix

# Display as DataFrame -->  Converts the matrix into a Pandas DataFrame
df = pd.DataFrame(matrix, index=vocabulary, columns=vocabulary)  # Rows = current wordsColumns = neighboring words


print(df)
```

### Exercise 4: Detect Data Sparsity
* Objective: Detect word pairs that never occur together using the Word Co-occurrence Matrix created in Exercise 3.

```python
import pandas as pd

# Input sentence
sentence = "the cat sat on the mat"
print("Input sentence: ", sentence)

# Split into words
words = sentence.split()

# Build vocabulary
vocabulary = sorted(set(words))  # Removes duplicate words and sorts the words alphabetically

# Create word-to-index mapping
word_index = {word: i for i, word in enumerate(vocabulary)}   # Creates a dictionary where every word gets an index

# Create empty co-occurrence matrix --> Creates a matrix initialized with zeros
matrix = [[0 for _ in vocabulary] for _ in vocabulary]   # Creates a 5 × 5 matrix filled with zeros

# Window size = 1
for i, word in enumerate(words):   # Loops through every word in the sentence

    if i > 0:   # Checks whether the current word(i) has a left neighbor
        left = words[i - 1]  # Gets the left neighboring word
        matrix[word_index[word]][word_index[left]] += 1   # Increases the count in the matrix

    if i < len(words) - 1:   # Checks whether the current word has a right neighbor
        right = words[i + 1]   # Gets the right neighboring word
        matrix[word_index[word]][word_index[right]] += 1   # Updates the matrix

# Display Co-occurrence Matrix --> Converts the matrix into a Pandas DataFrame
df = pd.DataFrame(matrix, index=vocabulary, columns=vocabulary)  # Rows = current words and Columns = neighboring words

print("Co-occurrence Matrix:\n")
print(df)

print("\nWord pairs that never occurred together:\n")

# Find zero-valued cells (excluding diagonal)
for i in range(len(vocabulary)):       # Loops through every row of the matrix
    for j in range(len(vocabulary)):    # Loops through every column, check every cell of the matrix
        if i != j and matrix[i][j] == 0: # Detect Sparsity, the two words never occurred together, A word is not compared with itself
            print(f"({vocabulary[i]}, {vocabulary[j]}) -> Never occurred")
```

### Exercise 5: Implement Laplace Smoothing
* Objective: Calculate word probabilities using Laplace Smoothing (Add-One Smoothing).

```python
# Input word counts --> A dictionary stores each word and its frequency
word_counts = {
    "cat": 4,
    "dog": 3,
    "bird": 1
}

# Compute total words --> Adds all word frequencies
total_words = sum(word_counts.values())
print("Total Words =", total_words)

# Compute vocabulary size --> Counts the number of unique words.
vocabulary_size = len(word_counts)
print("Vocabulary Size =", vocabulary_size)


print("\nLaplace Smoothed Probabilities:\n")
# Apply Laplace Smoothing
for word, count in word_counts.items():   # items() returns both the word and its count
    probability = (count + 1) / (total_words + vocabulary_size)   # Adds 1 to each word count before calculating the probability.
    print(f"{word} : {probability:.4f}")   # Displays the probability rounded to four decimal places.
```

### Exercise 6: Simple Keyboard Prediction
* Objective: Predict the most common next word based on the given training corpus using a Bigram Language Model.

```python
# Training corpus
corpus = [
    "I like NLP",
    "I like Python",
    "I like coffee",
    "I love NLP"
]

# Dictionary to store bigrams
bigrams = {}   # Creates an empty dictionary called bigrams

# Build bigram counts
for sentence in corpus:   # Reads one sentence at a time
    words = sentence.split()  # Splits each sentence into individual words

    for i in range(len(words) - 1):   # Loops through the words except the last one
        current = words[i]    # Stores the current word
        next_word = words[i + 1]  # Stores the next word

        if current not in bigrams:  # Checks whether the current word already exists in the dictionary
            bigrams[current] = {}   # Creates an empty dictionary for that word

        if next_word not in bigrams[current]: # Checks whether the next word has already seen after the current word
            bigrams[current][next_word] = 1  # Adds the next word with a count of 1
        else:
            bigrams[current][next_word] += 1  # If the same bigram appears again, increase its count by 1

# Input
current_word = "like"

# Display predictions
print("Suggestions:")

if current_word in bigrams:
    # Sort by highest frequency
    predictions = sorted(bigrams[current_word].items(),
                         key=lambda x: x[1],
                         reverse=True)

    for word, count in predictions[:3]:  # Takes the top 3 predictions
        print(word)
else:
    print("No prediction found")
```

### Exercise 7: Calculate TF-IDF
* Objective: Compute Term Frequency (TF), Document Frequency (DF), Inverse Document Frequency (IDF), and TF-IDF scores for the given corpus.

```python
import math  # It is used for the log10() function while calculating IDF

# Documents
documents = [           # Creates a list containing 3 documents
    "I like NLP",
    "I like AI",
    "AI is amazing"
]                                   

# ---------- TF ----------
tf = []   # Creates an empty list to store the TF of each document

for doc in documents:    # Loop one document at a time.
    words = doc.split()  # Splits the document into words
    total = len(words)   # Counts the total number of words

    freq = {}    # Creates an empty dictionary to store TF values
    for word in words:  # Loop every word in the document
        freq[word] = words.count(word) / total     # Formula TF = Number of times word appears / Total words in document

    tf.append(freq)

print("Term Frequency")
for i in range(len(tf)):
    print("Document", i + 1, ":", tf[i])

print("=" * 50)

# ---------- DF ----------
df = {}

for doc in documents:
    for word in set(doc.split()):  # set() removes duplicate words
        if word in df:  # Checks whether the word already exists
            df[word] += 1 # increase its document count
        else:
            df[word] = 1  # If no, add it with count 1

print("Document Frequency")
print(df)

print("=" * 50)

# ---------- IDF ----------
idf = {}         # Creates an empty dictionary
N = len(documents)  # Counts the number of documents

for word in df:
    idf[word] = math.log10(N / df[word])   # IDF Formula

print("Inverse Document Frequency")
print(idf)

print("=" * 50)

# ---------- TF-IDF ----------
print("TF-IDF")
for i in range(len(tf)):
    print("\nDocument", i + 1)

    for word in tf[i]:
        print(word, ":", round(tf[i][word] * idf[word], 4))   # TF-IDF=TF×IDF
```

### Exercise 8: Dense Vector Similarity
* Dense vector similarity measures how close or related two high-dimensional mathematical representations (dense vectors) are in a continuous embedding space.
* Objective: Compute cosine similarity between two dense vectors.
* 
* Tasks:
    * Compute dot product.
    * Compute vector magnitudes.
    * Calculate cosine similarity.

```python
import math

# Dense vectors
cat = [0.4, 0.2, 0.8]
dog = [0.5, 0.1, 0.7]

# Step 1: Compute dot product
dot_product = 0
for i in range(len(cat)):
    dot_product += cat[i] * dog[i]

# Step 2: Compute magnitudes
cat_magnitude = math.sqrt(sum(x * x for x in cat))
dog_magnitude = math.sqrt(sum(x * x for x in dog))

# Step 3: Compute cosine similarity
cosine_similarity = dot_product / (cat_magnitude * dog_magnitude)

# Output
print("Dot Product =", dot_product)
print("Magnitude of Cat =", round(cat_magnitude, 3))
print("Magnitude of Dog =", round(dog_magnitude, 3))
print("Cosine Similarity =", round(cosine_similarity, 2))
```

### Exercise 9: Find Similar Words
* Objective: Find the word most similar to a given word.
* Tasks
    * Compare the target vector with all others.
    * Compute cosine similarity.
    * Print the most similar word.
    * Formula: 

```python
import math

# Word embeddings --> Creates a dictionary called embeddings
embeddings = {
    "cat": [0.4, 0.2, 0.8],
    "dog": [0.5, 0.1, 0.7],
    "car": [-0.7, 0.8, -0.2],
    "lion": [0.45, 0.25, 0.82]
}

target = "cat"  # Compare all other words with cat

best_word = ""  # Creates an empty string, It will store the word that is most similar to cat
best_similarity = -1   # -1 → Exactly opposite direction, cosine similarity ranges from -1 to 1

for word in embeddings:   # Loops through every word in the dictionary
    if word != target:    # Skip comparing "cat" with itself
        a = embeddings[target]   # Gets the vector of "cat"
        b = embeddings[word]     # Gets the vector of the current word

        # Dot product
        dot = a[0]*b[0] + a[1]*b[1] + a[2]*b[2]  # Calculates the dot product, (0.4×0.5) + (0.2×0.1) + (0.8×0.7)

        # Magnitudes
        mag1 = math.sqrt(a[0]**2 + a[1]**2 + a[2]**2)  # Finds the magnitude (length) of the cat vector
        mag2 = math.sqrt(b[0]**2 + b[1]**2 + b[2]**2)  # Finds the magnitude (length) of the current word vector

        # Cosine similarity
        similarity = dot / (mag1 * mag2) # Calculates the cosine similarity

        if similarity > best_similarity:  # Check current similarity is greater than the previous best
            best_similarity = similarity  # Updates the highest similarity
            best_word = word    # Stores the word with the highest similarity

print("Most similar to", target)  # Most similar to cat
print(best_word)  # Most similar word
print("Similarity =", round(best_similarity, 2))  # Similarity rounded to 2 decimal places
```

### Exercise 10 (Assignment) : Mini NLP Pipeline
* Objective: Combine multiple concepts into one application.
* Tasks:
    1. Build the vocabulary.
    2. Count word frequencies.
    3. Compute TF.
    4. Build the co-occurrence matrix.
    5. Detect sparse entries.
    6. Apply Laplace smoothing to unigram probabilities.
    7. Predict the next word after a user-entered word.
    8. Display the top three most frequent words.

```python
from collections import Counter  # Counter class is used to count how many times each word appears

# Corpus --> Stores the sentences in a list
corpus = [
    "I love NLP",
    "I love AI",
    "AI loves Python",
    "Python loves data"
]

# 1. Build vocabulary
words = []  # Creates an empty list which store all words from every sentence
for sentence in corpus:  # Loops through each sentence in the corpus
    words.extend(sentence.split())  # Split Sentence into Words, Adds these words into the list

vocab = sorted(set(words))  # Removes duplicate words, Sorts the words alphabetically
print("Vocabulary Size :", len(vocab))  # Vocabulary size

# 2. Count word frequencies
freq = Counter(words)   # Counts how many times each word appears

# 3. Compute TF
print("\nTerm Frequency")
total = len(words)  # Counts total word length
for word in freq:   # Loops through every unique word
    print(word, ":", round(freq[word] / total, 2))  # Compute TF = Frequency / Total Words

# 4. Build co-occurrence matrix
co = {}  # Creates an empty dictionary, It will store word pairs

for sentence in corpus:  # Reads one sentence at a time
    w = sentence.split()  # Split sentence
    for i in range(len(w) - 1):  # Moves through adjacent words
        pair = (w[i], w[i + 1])  # Creates a tuple pair
        co[pair] = co.get(pair, 0) + 1  # Count Pair Frequency, if pair already exists increase by 1 otherwise starts with 0

# 5. Detect sparse entries
print("\nSparse Pairs")
for w1 in vocab:     # Checks every possible combination
    for w2 in vocab:   
        if (w1, w2) not in co:  # Find missing pairs
            print("(", w1, ",", w2, ")", sep="")  # Print sparse pair

# 6. Laplace smoothing
print("\nLaplace Probabilities")
V = len(vocab)   # Vocabulary size
for word in vocab:
    p = (freq[word] + 1) / (total + V)  # Calculates probability for every word, Laplace Smoothing=(Frequency + 1)/(Total Words + Vocabulary Size)
    print(word, ":", round(p, 3))  # Round upto 3 decimal places

# 7. Predict next word
target = input("\nEnter a word: ")  # Take user unput

print("Prediction after", target)
found = False  # Assumes no prediction is found
for (w1, w2), count in co.items():  # Reads every stored pair
    if w1 == target:  # Checks whether the first word equals the user's input
        print(w2)   # Print next word
        found = True   # Marks that at least one prediction exists

if not found:   # If the word doesn't exist
    print("No prediction")

# 8. Top three frequent words
print("\nTop 3 Words")
for word, count in freq.most_common(3):  # Returns the three highest-frequency words(most frequent 3 word)
    print(word, ":", count)  # Print word and it's count
```
